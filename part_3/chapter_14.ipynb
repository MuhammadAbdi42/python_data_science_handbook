## Chapter 14 - Data Indexing and Selection

```python
import pandas as pd
import numpy as np
```

### Data Selection in Series

#### Series as Dictionary

- Like a dictionary, the Series object provides a mapping from a collection of keys to a collection of values:

```python
data = pd.Series(np.linspace(0.25, 1, 4), index = ['a', 'b', 'c', 'd'])
print(data['b'])
```

```python
# Using dictionary-like Python expressions
print('a' in data)
print(data.keys())
print(list(data.items()))
```

```python
# Modifying
data['e'] = 1.25
print(data)
```

#### Series as One-Dimensional Array

```python
# Slicing by explicit index
print(data['a':'c'])

# Slicing by implicit integer index
print(data[0:2])

# Masking
print(data[(data > 0.3) & (data < 0.8)])

# Fancy indexing
print(data[['a', 'e']])
```

- **IMPORTANT:** Notice that when slicing with an explicit index (e.g., `data['a':'c']`), the final index is included in the slice, while when slicing with an implicit index (e.g., `data[0:2]`), the final index is excluded from the slice.

#### Indexers: `loc` and `iloc`

- If your Series has an explicit integer index, an indexing operation such as `data[1]` will use the explicit indices, while a slicing operation like `data[1:3]` will use the implicit Python-style indices.
- Because of this potential confusion in the case of integer indexes, Pandas provides some special indexer attributes that explicitly expose certain indexing schemes. These are not functional methods, but attributes that expose a particular slicing interface to the data in the Series.

```python
data = pd.Series(['a', 'b', 'c'], index = [1, 3, 5])
print(data)

print(data[1])

print(data[1:3])
```

First, the `loc` attribute allows indexing and slicing that always references the explicit index:

```python
data.loc[1]
```

- The `iloc` attribute allows indexing and slicing that always references the implicit Python-style index:

```python
data.iloc[1]
```

### Data Selection in DataFrames

#### DataFrame as Dictionary

```python
area = pd.Series({'California': 423967, 'Texas': 695662,  'Florida': 170312, 'New York': 141297, 'Pennsylvania': 119280})
pop = pd.Series({'California': 39538223, 'Texas': 29145505,  'Florida': 21538187, 'New York': 20201249, 'Pennsylvania': 13002700})
data = pd.DataFrame({'area':area, 'pop':pop})

print(data)
```

- The individual Series that make up the columns of the DataFrame can be accessed via dictionary-style indexing of the column name.
- Equivalently, we can use attribute-style access with column names that are strings.

```python
print(data['area'])
print(data.area)
```

- Like with the `Series` objects discussed earlier, this dictionary-style syntax can also be used to modify the object, in this case adding a new column:

```python
data['density'] = data['pop'] / data['area']
print(data)
```

#### DataFrame as Two-Dimensional Array

- As mentioned previously, we can also view the `DataFrame` as an enhanced twodimensional array.

```python
data.values
```

- We can transpose the full DataFrame to swap rows and columns.

```python
data.T
```

- Using the `iloc` indexer, we can index the underlying array as if it were a simple NumPy array (using the implicit Pythonstyle index), but the DataFrame index and column labels are maintained in the result:

```python
data.iloc[:3, :2]
```

- Similarly, using the `loc` indexer we can index the underlying data in an array-like style but using the explicit index and column names:

```python
data.loc[:'Florida', :'pop']
```

- In the loc indexer we can combine masking and fancy indexing as follows:

```python
data.loc[data.density > 120, ['pop', 'density']]
```

- Any of these indexing conventions may also be used to set or modify values; this is done in the standard way that you might be accustomed to from working with NumPy:

```python
data.iloc[0, 2] = 90
print(data)
```
